#### Peaks calculation

```python
def find_spectrum_peaks_adaptive(
    wavenumber, spectrum,
    base_prom: float = 0.03,
    base_height: float | None = 0.05,    # поставьте None, если хотите больше чувствительности
    k_prom: float = 3.0,                 # оставлены для совместимости, в логике не используются
    k_height: float = 2.0,               # оставлены для совместимости, в логике не используются
    min_width_cm1: float = 8.0,
    min_distance_cm1: float = 10.0,
):
    x = np.asarray(wavenumber, dtype=float)
    y = np.asarray(spectrum, dtype=float)

    # шаг по оси в см⁻¹ (median — чуть устойчивей к выбросам)
    dx = np.median(np.abs(np.diff(x)))
    if not np.isfinite(dx) or dx <= 0:
        dx = np.mean(np.abs(np.diff(x))) or 1.0

    # перевод ограничений из см⁻¹ в точки
    to_pts = lambda cm1: max(1, int(round(cm1 / dx)))
    min_width_pts = to_pts(min_width_cm1)
    min_distance_pts = to_pts(min_distance_cm1)

    # Поиск пиков
    kwargs = dict(prominence=base_prom, width=min_width_pts, distance=min_distance_pts)
    if base_height is not None:
        kwargs["height"] = base_height
    peaks, props = find_peaks(y, **kwargs)

    # Робастная оценка шума: sigma ≈ 1.4826 * MAD
    y_med = np.median(y)
    mad = np.median(np.abs(y - y_med))
    sigma = 1.4826 * mad if mad > 0 else (np.std(y) or 1e-6)

    snr = props.get("peak_heights", np.ones_like(peaks)) / (sigma if sigma > 0 else 1e-6)
    edge_mask = (peaks == 0) | (peaks == len(y) - 1)
    widths_cm1 = props.get("widths", np.zeros_like(peaks, dtype=float)) * abs(dx)

    props_out = {
        "snr": snr,
        "edge_peak": edge_mask,
        "widths_cm1": widths_cm1,
    }
    return x[peaks], y[peaks], props_out
```

```python
def peaks_dict_to_long(peak_data):
    rows = []
    for sid, pdata in peak_data.items():
        for pos, inten in zip(pdata["positions"], pdata["intensities"]):
            rows.append({
                "sample_id": sid,
                "pos": float(pos),
                "intensity": float(inten)
            })
    df = pd.DataFrame(rows).sort_values(by=["sample_id", "pos"])
    return df
```

#### Function for specra with peaks visualization

```python
def visualize_grouped_spectra(
    grouped_data: dict[str, pd.DataFrame],
    peak_data: dict,
    groups: list[str],
    type_color_dict: dict[str, str] | None = None,
    plot_style: dict | None = None,
    plot_params: dict | None = None,
    path_to_save: str | None = None,
    x_inverted: bool = True,
):
    plot_style = (plot_style or PLOT_STYLE).copy()
    plot_params = (plot_params or PLOT_PARAMS).copy()
    type_color_dict = type_color_dict or {g: "blue" for g in groups}

    ncols = len(groups)
    nrows = max(len(grouped_data[g]) for g in groups)
    figsize = (ncols * plot_params["FIGSIZE_SCALE"][0],
               nrows * plot_params["FIGSIZE_SCALE"][1])

    fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=figsize, dpi=300)
    axes = np.atleast_2d(axes)

    for col, group in tqdm(list(enumerate(groups)), total=len(groups)):
        group_df = grouped_data[group]
        color = type_color_dict.get(group, "blue")
        x_vals = group_df.columns.astype(float).values  # одна ось X на всю группу

        for row, sample_id in enumerate(group_df.index):
            ax = axes[row, col]
            y_vals = group_df.loc[sample_id].values

            # Автомасштаб по Y
            ymin, ymax = y_vals.min() * 0.95, y_vals.max() * 1.05
            ax.set_ylim(ymin, ymax)
            ax.plot(x_vals, y_vals, color=color, linewidth=plot_style["LINEWIDTH"])

            # Пики
            info = peak_data.get(sample_id, {})
            x_pk = info.get("positions", [])
            y_pk = info.get("intensities", [])
            snr_arr = info.get("snr", np.full(len(x_pk), np.nan, dtype=float))
            edge_arr = info.get("edge_peak", np.zeros(len(x_pk), dtype=bool))

            for xp, yp, snr_v, is_edge in zip(x_pk, y_pk, snr_arr, edge_arr):
                if is_edge:
                    mcolor = "gray"
                elif np.isfinite(snr_v) and snr_v < 5:
                    mcolor = "orange"
                else:
                    mcolor = "red"
                ax.scatter(xp, yp, s=plot_style["PEAK_SIZE"], color=mcolor, edgecolor="k", zorder=3)

            if row == 0:
                ax.set_title(f"{group}", fontsize=plot_style["TITLE_FONTSIZE"], fontstyle="italic")

            ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=plot_style["X_NBINS"]))
            ax.yaxis.set_major_locator(ticker.MaxNLocator(nbins=plot_style["Y_NBINS"]))
            ax.tick_params(axis="x", labelsize=plot_style["TICK_FONTSIZE"])
            ax.tick_params(axis="y", labelsize=plot_style["TICK_FONTSIZE"])

            # Подписи (оставим на каждой оси — на научных панелях так бывает удобнее)
            ax.set_xlabel(plot_params["X_LABEL"], fontsize=plot_style["X_LABEL_FONT_SIZE"],
                          labelpad=plot_style["X_LABELPAD"])
            ax.set_ylabel(plot_params["Y_LABEL"], fontsize=plot_style["Y_LABEL_FONT_SIZE"],
                          rotation=0, labelpad=plot_style["Y_LABELPAD"])

            if x_inverted:
                ax.invert_xaxis()

    # Скрываем пустые оси, если в колонке меньше рядов, чем nrows
    for col, group in enumerate(groups):
        n_in_col = len(grouped_data[group])
        for row in range(n_in_col, nrows):
            axes[row, col].axis("off")

    plt.tight_layout()
    if path_to_save is not None:
        plt.savefig(path_to_save, dpi=600)
    plt.show()
```

#### Function for peaks aggregation

```python
def consolidate_peaks_dbscan(
    peaks_long: pd.DataFrame,
    eps_cm1: float = 32.0,                       # DBSCAN radius (ε) in cm⁻¹
    min_samples: int = 1,                        # 1 => any point can start/extend a cluster
    centroid: str = "mean",                      # 'median' | 'mean' | 'weighted_mean'
    per_sample: str = "weighted_mean_intensity", # 'nearest' | 'max_intensity' | 'weighted_mean_intensity'
    keep_noise: bool = True,                     # include DBSCAN noise (label == -1) as separate "clusters"
    name_decimals: int = 1                       # kept for compatibility; not used when centroids are integer-rounded
):
    req = {"sample_id", "pos", "intensity"}
    if peaks_long.empty:
        return pd.DataFrame(), pd.DataFrame()
    if not req.issubset(peaks_long.columns):
        raise ValueError(f"peaks_long must contain columns {req}")

    # Keep only the required columns, drop NaNs, sort once by position
    df = (peaks_long[["sample_id", "pos", "intensity"]]
          .dropna()
          .sort_values("pos", kind="mergesort")  # stable sort preserves potential group order
          .reset_index(drop=True))

    # --- 1) DBSCAN over 1D positions ---
    X = df["pos"].to_numpy(dtype=float).reshape(-1, 1)
    db = DBSCAN(eps=eps_cm1, min_samples=min_samples, metric="euclidean")
    df["cluster"] = db.fit_predict(X)  # -1 means noise

    if not keep_noise:
        df = df[df["cluster"] != -1].copy()
        if df.empty:
            return pd.DataFrame(), pd.DataFrame()

    # --- 2) Cluster centroids (numeric anchors for columns), then ROUND TO INTEGERS ---
    def _centroid(g: pd.DataFrame) -> float:
        # Compute cluster anchor before rounding
        if centroid == "weighted_mean":
            w = g["intensity"].to_numpy()
            wsum = w.sum()
            return float(np.average(g["pos"], weights=w)) if wsum > 0 else float(g["pos"].median())
        elif centroid == "median":
            return float(g["pos"].median())
        else:
            return float(g["pos"].mean())

    centroids_df = (df.groupby("cluster", sort=True, observed=True)
                      .apply(_centroid)
                      .rename("centroid")
                      .reset_index())

    # Round centroids to nearest integer for stable, integer wavenumbers
    centroids_df["centroid"] = centroids_df["centroid"].round(0).astype(int)

    # Stable, numeric ordering by rounded centroid
    centroids_df = centroids_df.sort_values("centroid", kind="mergesort").reset_index(drop=True)

    # Quick lookup: cluster -> (integer) centroid
    cl_to_cen = dict(zip(centroids_df["cluster"].to_numpy(), centroids_df["centroid"].to_numpy()))

    # --- 3) Pick exactly one representative per sample within each cluster ---
    def _pick_rows_in_cluster(g: pd.DataFrame, cen_int: int) -> pd.DataFrame:
        if per_sample == "max_intensity":
            # index of max intensity per sample
            idx = g.groupby("sample_id", observed=True)["intensity"].idxmax()
            chosen = g.loc[idx, ["sample_id", "pos", "intensity"]]
            return chosen

        if per_sample == "nearest":
            # distance to (integer) centroid, then argmin per sample
            dist = (g["pos"] - cen_int).abs()
            g2 = g.loc[dist.index]
            g2 = g2.assign(_dist=dist.to_numpy())
            chosen = (g2.sort_values(["sample_id", "_dist"], kind="mergesort")
                        .drop_duplicates("sample_id", keep="first")
                        .drop(columns="_dist"))[["sample_id", "pos", "intensity"]]
            return chosen

        # 'weighted_mean_intensity': collapse multiple peaks of the same sample into one
        # position = weighted average by intensity; intensity = intensity weighted by itself (power-biased on purpose)
        def _combine(h: pd.DataFrame) -> pd.Series:
            I = h["intensity"].to_numpy()
            s = I.sum()
            if s == 0.0:
                # fallback: nearest to the integer centroid
                ridx = (h["pos"] - cen_int).abs().idxmin()
                return pd.Series({"pos": h.at[ridx, "pos"], "intensity": 0.0}, dtype=float)
            pos_w = float(np.average(h["pos"], weights=I))
            inten_w = float(np.average(I, weights=I))
            return pd.Series({"pos": pos_w, "intensity": inten_w}, dtype=float)

        chosen = (g.groupby("sample_id", observed=True, as_index=False)
                    .apply(_combine)
                    .reset_index(drop=True))
        return chosen[["sample_id", "pos", "intensity"]]

    picked_parts = []
    for cl, g in df.groupby("cluster", sort=False, observed=True):
        cen_int = int(cl_to_cen[cl])
        chosen = _pick_rows_in_cluster(g, cen_int)
        if not chosen.empty:
            chosen = chosen.copy()
            chosen["centroid_cm1"] = cen_int  # store integer-rounded centroid
            picked_parts.append(chosen)

    if not picked_parts:
        return pd.DataFrame(), pd.DataFrame()

    picked_df = pd.concat(picked_parts, ignore_index=True)

    # --- 4) Build integer column names and pivot to wide matrix ---
    def _mk_name_int(x_int: int) -> str:
        # Integer wavenumber column name, e.g. $I_{3435}$
        return f"$I_{{{int(x_int)}}}$"

    unique_cen = (picked_df[["centroid_cm1"]]
                  .drop_duplicates(ignore_index=True)
                  .sort_values("centroid_cm1", kind="mergesort")
                  .reset_index(drop=True))
    unique_cen["column"] = unique_cen["centroid_cm1"].apply(_mk_name_int)

    picked_df = picked_df.merge(unique_cen, on="centroid_cm1", how="left", copy=False)

    # Pivot to wide
    mat = picked_df.pivot(index="sample_id", columns="column", values="intensity")

    # --- 5) Build info summary (coverage + raw cluster stats) ---
    info = (picked_df.groupby(["centroid_cm1", "column"], observed=True)
                  .agg(n_picked=("intensity", "count"))
                  .reset_index())

    # Ensure info sorted by centroid (ascending wavenumber)
    info = info.sort_values("centroid_cm1", kind="mergesort").reset_index(drop=True)

    # Align mat columns strictly to sorted info order (ascending wavenumber)
    mat = mat.reindex(columns=info["column"].tolist())

    # Coverage (use aligned mat)
    info["n_non_nan"] = mat.notna().sum(axis=0).to_numpy()
    info["n_nan"] = mat.isna().sum(axis=0).to_numpy()
    info["non_nan_ratio"] = info["n_non_nan"] / (info["n_non_nan"] + info["n_nan"])

    # Raw cluster stats based on original assignments (before per-sample consolidation)
    raw_stats = (df.groupby("cluster", observed=True)
                   .agg(
                       span_cm1=("pos", lambda x: float(x.max() - x.min()) if len(x) else 0.0),
                       n_points=("pos", "size"),
                       n_unique_samples=("sample_id", "nunique"),
                   )
                   .reset_index())

    # Map cluster by integer centroid back into info
    centroids_df_ren = centroids_df.rename(columns={"cluster": "cluster_id"})
    info = (info.merge(centroids_df_ren, left_on="centroid_cm1", right_on="centroid", how="left")
                .merge(raw_stats, left_on="cluster_id", right_on="cluster", how="left")
                .drop(columns=["centroid", "cluster"])
                .sort_values("centroid_cm1", kind="mergesort")
                .reset_index(drop=True))

    return mat.astype(float), info
```

#### Filling *NaN*-values

```python
def fill_missing_with_nearest(peaks_df, grouped_data, info):
    filled_df = peaks_df.copy()

    sample_spectra = {}
    for group_name, df_group in grouped_data.items():
        wn = df_group.columns.astype(float).values
        for sample_id in df_group.index:
            sample_spectra[sample_id] = (wn, df_group.loc[sample_id].values)

    for _, row in info.iterrows():
        centroid = row["centroid_cm1"]
        col_name = row["column"]

        if col_name not in filled_df.columns:
            continue

        for sample_id in filled_df.index[filled_df[col_name].isna()]:
            wn, intensities = sample_spectra[sample_id]
            idx = np.argmin(np.abs(wn - centroid))
            filled_df.at[sample_id, col_name] = intensities[idx]

    return filled_df
```

#### Processing the *test* part of dataset

```python
def detect_peaks_for_df(data_df: pd.DataFrame, detect_fn) -> pd.DataFrame:
    rows = []
    wn_axis = data_df.columns.astype(float).values
    for sid, y in data_df.iterrows():
        pos, inten, _ = detect_fn(wn_axis, y.values)
        if len(pos):
            rows.extend({"sample_id": sid, "pos": float(p), "intensity": float(i)}
                        for p, i in zip(pos, inten))
    return pd.DataFrame(rows, columns=["sample_id", "pos", "intensity"])


def _column_to_centroid_map(info: pd.DataFrame) -> dict:
    return dict(zip(info["column"], info["centroid_cm1"]))
```

```python
def build_test_from_train_columns_dbscan(
    data_test: pd.DataFrame,
    info_train_subset: pd.DataFrame,          # 'info' filtered to train_columns
    train_columns: list[str],                  # columns of peaks_..._df_train after dropping rare features
    peak_tolerance: float,                     # use SAME value as eps_cm1 used on train
    detect_fn,                                 # the same detector used on train (e.g., find_spectrum_peaks_adaptive)
    per_sample: str = "weighted_mean_intensity" # 'weighted_mean_intensity' | 'nearest' | 'max_intensity'
) -> pd.DataFrame:
    # 1) keep only expected columns and preserve train order
    info_f = info_train_subset[info_train_subset["column"].isin(train_columns)].copy()
    info_f = info_f.set_index("column").loc[train_columns].reset_index()
    col2cent = dict(zip(info_f["column"], info_f["centroid_cm1"]))  # integer centroids

    # 2) detect all local peaks on test, then pre-split by sample for fast access
    peaks_long_test = detect_peaks_for_df(data_test, detect_fn)
    peaks_by_sample = {
        sid: sdf[["pos", "intensity"]].reset_index(drop=True)
        for sid, sdf in peaks_long_test.groupby("sample_id")
    }

    # 3) empty test matrix with fixed train columns
    mat_test = pd.DataFrame(index=data_test.index, columns=train_columns, dtype=float)

    # 4) fill per sample
    for sid in mat_test.index:
        sdf = peaks_by_sample.get(sid)
        if sdf is None or sdf.empty:
            continue

        pos_arr = sdf["pos"].to_numpy(float)
        inten_arr = sdf["intensity"].to_numpy(float)
        # Spectra assumed normalized; intensities are non-negative by your data contract
        I_nonneg = inten_arr  # no clipping needed

        for col in train_columns:
            c = col2cent[col]  # integer centroid
            # window ± peak_tolerance
            mask = np.abs(pos_arr - c) <= peak_tolerance
            if not np.any(mask):
                # leave NaN; may be filled with baseline later
                continue

            if per_sample == "weighted_mean_intensity":
                # intensity (goes to matrix): power-biased average = sum(I^2) / sum(I), I >= 0
                I = I_nonneg[mask]
                ISum = I.sum()
                if ISum <= 0:
                    # fallback: pick the maximum absolute intensity within window
                    j = np.argmax(np.abs(inten_arr[mask]))
                    mat_test.at[sid, col] = float(inten_arr[mask][j])
                else:
                    I_avg = float((I ** 2).sum() / ISum)
                    mat_test.at[sid, col] = I_avg

            elif per_sample == "max_intensity":
                j = np.argmax(inten_arr[mask])
                mat_test.at[sid, col] = float(inten_arr[mask][j])

            else:  # 'nearest'
                j = np.argmin(np.abs(pos_arr[mask] - c))
                mat_test.at[sid, col] = float(inten_arr[mask][j])

    return mat_test
```

```python
def fill_test_nans_with_nearest_baseline(
    mat_test: pd.DataFrame,
    data_test: pd.DataFrame,
    info_train_subset: pd.DataFrame,
) -> pd.DataFrame:
    filled = mat_test.copy()
    col2cent = _column_to_centroid_map(info_train_subset[info_train_subset["column"].isin(filled.columns)])

    wn_axis = data_test.columns.astype(float).values
    y_by_sample = {sid: data_test.loc[sid].values for sid in data_test.index}

    for col in filled.columns:
        c = col2cent[col]  # integer centroid
        # single nearest index per column speeds up inner loop
        nearest_idx = int(np.argmin(np.abs(wn_axis - c)))
        for sid in filled.index[filled[col].isna()]:
            y = y_by_sample[sid]
            filled.at[sid, col] = float(y[nearest_idx])
    return filled
```

### Additional features calculation

#### Base function

```python
def compute_interval_and_peak_region_features(spectra_data, peaks_df, intervals):
    df = pd.DataFrame(index=spectra_data.index)
    wn = spectra_data.columns.astype(float)

    for a, b in intervals:
        low, high = min(a, b), max(a, b)

        # Mean intensity in intervals
        mean_name = f"$\\overline{{I}}_{{{high}\\text{{–}}{low}}}$"
        mask = (wn >= low) & (wn <= high)
        df[mean_name] = spectra_data.loc[:, mask].mean(axis=1)

        # Sum of intensities of peaks in intervals — using Σ
        sum_name = f"$\\Sigma I_{{p,\\ {high}\\text{{–}}{low}}}$"
        selected_cols = [
            col for col in peaks_df.columns
            if col.startswith("$I_{") and low <= int(col.split('{')[1].split('}')[0]) <= high
        ]
        df[sum_name] = peaks_df[selected_cols].sum(axis=1)

    return df
```

```python
def compute_interval_and_peak_region_features(spectra_data, peaks_df, intervals, windows=(25, ), compute_peak_region_mean=True):
    df = pd.DataFrame(index=spectra_data.index)
    wn = spectra_data.columns.astype(float)

    for a, b in intervals:
        low, high = min(a, b), max(a, b)

        # Mean intensity in intervals
        mean_name = f"$\\overline{{I}}_{{{high}\\text{{–}}{low}}}$"
        mask = (wn >= low) & (wn <= high)
        df[mean_name] = spectra_data.loc[:, mask].mean(axis=1)

        # Sum of intensities of peaks in intervals — using Σ
        sum_name = f"$\\Sigma I_{{p,\\ {high}\\text{{–}}{low}}}$"
        selected_cols = [
            col for col in peaks_df.columns
            if col.startswith("$I_{") and low <= int(col.split('{')[1].split('}')[0]) <= high
        ]
        df[sum_name] = peaks_df[selected_cols].sum(axis=1)

    if compute_peak_region_mean == True:
        # Mean of intensity in range of peak maximum ± window
        for sample_id in spectra_data.index:
            row = peaks_df.loc[sample_id].dropna()
            for col in row.index:
                if col.startswith("$I_{"):
                    try:
                        center = int(col.split('{')[1].split('}')[0])
                    except ValueError:
                        continue
    
                    for w in windows:
                        left, right = center - w, center + w
                        mask = (wn >= left) & (wn <= right)
                        if mask.any():
                            region_mean = spectra_data.loc[sample_id, wn[mask]].mean()
                            feat_name = f"$\\overline{{I}}_{{{center} \\pm {w}}}$"
                            df.at[sample_id, feat_name] = region_mean
    else:
        pass

    return df
```
